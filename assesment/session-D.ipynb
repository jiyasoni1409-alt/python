```python
# Food Delivery Bill Calculator

items = []


while True:
    try:
        n = int(input("Enter number of food items: "))

        if n <= 0:
            print("Error: Number of items must be greater than 0.")
        else:
            break

    except ValueError:
        print("Error: Please enter a valid number.")



for i in range(n):

    name = input(f"Enter item {i + 1} name: ").strip()

    while name == "":
        print("Error: Item name cannot be empty.")
        name = input(f"Enter item {i + 1} name: ").strip()

    while True:
        try:
            price = float(input(f"Enter price of {name}: Rs "))

            if price < 0:
                print("Error: Price cannot be negative.")
            else:
                break

        except ValueError:
            print("Error: Please enter a valid numeric price.")

    items.append((name, price))



while True:
    try:
        order_count = int(input("Enter number of previous orders: "))

        if order_count < 0:
            print("Error: Order count cannot be negative.")
        else:
            break

    except ValueError:
        print("Error: Please enter a valid whole number.")

subtotal = 0

for item in items:
    subtotal += item[1]

gst = subtotal * 0.18



delivery_fee = 30

discount = 0

if order_count > 5:
    discount = subtotal * 0.10



final_amount = subtotal + gst + delivery_fee - discount



print("\n==========================================")
print("          FOOD DELIVERY RECEIPT")
print("==========================================")

for item in items:
    print(f"{item[0]:25} Rs {item[1]:.2f}")

print("------------------------------------------")
print(f"{'Subtotal':25} Rs {subtotal:.2f}")
print(f"{'GST (18%)':25} Rs {gst:.2f}")
print(f"{'Delivery Fee':25} Rs {delivery_fee:.2f}")

if discount > 0:
    print(f"{'Loyalty Discount (10%)':25} -Rs {discount:.2f}")
else:
    print(f"{'Loyalty Discount':25} Rs 0.00")

print("------------------------------------------")
print(f"{'Final Amount Payable':25} Rs {final_amount:.2f}")
print("==========================================")
```
