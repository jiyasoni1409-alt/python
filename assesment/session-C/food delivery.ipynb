```python
import json
import os


# ==============================
# Order Class
# ==============================

class Order:

    def __init__(self, order_id, customer_name, items, total_amount, status="Pending"):
        self.order_id = order_id
        self.customer_name = customer_name
        self.items = items
        self.total_amount = total_amount
        self.status = status

    # Convert object into dictionary for JSON
    def to_dict(self):
        return {
            "order_id": self.order_id,
            "customer_name": self.customer_name,
            "items": self.items,
            "total_amount": self.total_amount,
            "status": self.status
        }


# ==============================
# Load Orders from JSON
# ==============================

def load_orders():
    try:
        with open("orders.json", "r") as file:
            data = json.load(file)

            orders = []

            for order_data in data:
                order = Order(
                    order_data["order_id"],
                    order_data["customer_name"],
                    order_data["items"],
                    order_data["total_amount"],
                    order_data["status"]
                )

                orders.append(order)

            return orders

    except FileNotFoundError:
        print("orders.json not found. Starting with an empty order list.")
        return []

    except json.JSONDecodeError:
        print("Error: orders.json contains invalid data.")
        return []


# ==============================
# Save Orders to JSON
# ==============================

def save_orders(orders):

    data = []

    for order in orders:
        data.append(order.to_dict())

    with open("orders.json", "w") as file:
        json.dump(data, file, indent=4)


# ==============================
# Generate Order ID
# ==============================

def generate_order_id(orders):

    if len(orders) == 0:
        return "ORD001"

    numbers = []

    for order in orders:
        number = int(order.order_id.replace("ORD", ""))
        numbers.append(number)

    next_number = max(numbers) + 1

    return f"ORD{next_number:03d}"


# ==============================
# Place New Order
# ==============================

def place_order(orders):

    print("\n========== PLACE NEW ORDER ==========")

    # Customer name validation
    while True:
        customer_name = input("Enter customer name: ").strip()

        if customer_name == "":
            print("Error: Customer name cannot be empty.")
        else:
            break

    # Items
    while True:
        items_input = input("Enter items (comma-separated): ").strip()

        if items_input == "":
            print("Error: Please enter at least one item.")
        else:
            items = [item.strip() for item in items_input.split(",")]
            items = [item for item in items if item != ""]
            break

    # Amount validation
    while True:
        try:
            total_amount = float(input("Enter total amount (Rs): "))

            if total_amount < 0:
                print("Error: Amount cannot be negative.")
            else:
                break

        except ValueError:
            print("Error: Please enter a valid numeric amount.")

    # Generate ID
    order_id = generate_order_id(orders)

    # Create Order object
    new_order = Order(
        order_id,
        customer_name,
        items,
        total_amount
    )

    # Add order to list
    orders.append(new_order)

    # Save immediately
    save_orders(orders)

    print("\nOrder placed successfully!")
    print(f"Order ID: {order_id}")
    print(f"Customer: {customer_name}")
    print(f"Total Amount: Rs {total_amount:.2f}")
    print("Status: Pending")


# ==============================
# View All Orders
# ==============================

def view_all_orders(orders):

    print("\n================ ALL ORDERS ================")

    if len(orders) == 0:
        print("No orders found.")
        return

    print(
        f"{'Order ID':<12}"
        f"{'Customer Name':<20}"
        f"{'Items':<10}"
        f"{'Amount':<15}"
        f"{'Status'}"
    )

    print("-" * 70)

    for order in orders:

        status = order.status

        # Highlight Delivered orders
        if status.lower() == "delivered":
            status = f"*** {status} ***"

        print(
            f"{order.order_id:<12}"
            f"{order.customer_name:<20}"
            f"{len(order.items):<10}"
            f"Rs {order.total_amount:<12.2f}"
            f"{status}"
        )

    print("-" * 70)


# ==============================
# Search Order by ID
# ==============================

def search_order(orders):

    print("\n========== SEARCH ORDER ==========")

    order_id = input("Enter Order ID: ").strip().upper()

    found = False

    for order in orders:

        if order.order_id.upper() == order_id:

            print("\nOrder Found!")
            print("-----------------------------")
            print(f"Order ID      : {order.order_id}")
            print(f"Customer Name : {order.customer_name}")
            print(f"Items         : {', '.join(order.items)}")
            print(f"Item Count    : {len(order.items)}")
            print(f"Total Amount  : Rs {order.total_amount:.2f}")
            print(f"Status        : {order.status}")
            print("-----------------------------")

            found = True
            break

    if not found:
        print("Order not found.")


# ==============================
# Main Program
# ==============================

orders = load_orders()

while True:

    print("\n======================================")
    print("     FOOD DELIVERY ORDER SYSTEM")
    print("======================================")
    print("1. Place New Order")
    print("2. View All Orders")
    print("3. Search Order by ID")
    print("4. Exit")
    print("======================================")

    choice = input("Enter your choice: ").strip()

    if choice == "1":

        place_order(orders)

    elif choice == "2":

        view_all_orders(orders)

    elif choice == "3":

        search_order(orders)

    elif choice == "4":

        print("\nThank you for using Food Delivery Order System!")
        break

    else:

        print("Invalid choice. Please enter 1, 2, 3, or 4.")
```
