```python
# Restaurant Menu 

menu = {
    "Margherita Pizza": {"price": 250, "category": "Pizza"},
    "Veg Burger": {"price": 150, "category": "Burger"},
    "Pasta": {"price": 200, "category": "Italian"},
    "French Fries": {"price": 100, "category": "Snacks"},
    "Veg Sandwich": {"price": 120, "category": "Snacks"},
    "Paneer Tikka": {"price": 220, "category": "Starter"}
}


def view_menu():
    print("\n----- Restaurant Menu -----")
    print(f"{'No.':<5}{'Dish Name':<25}{'Price':<10}{'Category'}")
    print("-" * 55)

    for number, (dish, details) in enumerate(menu.items(), start=1):
        print(f"{number:<5}{dish:<25}Rs {details['price']:<7}{details['category']}")



def filter_by_category():
    category = input("Enter category: ").strip()

    print(f"\nItems in category: {category}")
    found = False

    for dish, details in menu.items():
        if details["category"].lower() == category.lower():
            print(f"{dish} - Rs {details['price']}")
            found = True

    if not found:
        print("No items found in this category.")


def search_dish():
    dish_name = input("Enter dish name: ").strip()

    found = False

    for dish, details in menu.items():
        if dish.lower() == dish_name.lower():
            print(f"\nDish: {dish}")
            print(f"Price: Rs {details['price']}")
            found = True
            break

    if not found:
        print("Dish not found.")


while True:
    print("\n===== Restaurant Menu Manager =====")
    print("1. View all items")
    print("2. Filter items by category")
    print("3. Search for a dish")
    print("0. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        view_menu()

    elif choice == "2":
        filter_by_category()

    elif choice == "3":
        search_dish()

    elif choice == "0":
        print("Thank you! Program ended.")
        break

    else:
        print("Invalid choice. Please try again.")
```
