```python
import json


def add_order():
    try:
    
        customer_name = input("Enter customer name: ")

        items_input = input("Enter items (comma-separated): ")
        items = [item.strip() for item in items_input.split(",")]

        total_amount = float(input("Enter total amount (Rs): "))

        order_status = input("Enter order status: ")

        
        try:
            with open("orders.json", "r") as file:
                orders = json.load(file)
        except FileNotFoundError:
            # If file does not exist, start with an empty list
            orders = []

        
        new_order = {
            "customer_name": customer_name,
            "items": items,
            "total_amount": total_amount,
            "order_status": order_status
        }

        
        orders.append(new_order)

        
        with open("orders.json", "w") as file:
            json.dump(orders, file, indent=4)

        print("\nOrder saved successfully!")

    except ValueError:
        print("Error: Total amount must be a number.")


# Function to view all orders
def view_orders():
    try:
        with open("orders.json", "r") as file:
            orders = json.load(file)

        if not orders:
            print("\nNo orders found.")
            return

        print("\n========== ORDER HISTORY ==========")

        for number, order in enumerate(orders, start=1):
            print(f"\nOrder {number}")
            print(f"Customer Name : {order['customer_name']}")
            print(f"Items         : {', '.join(order['items'])}")
            print(f"Total Amount  : Rs {order['total_amount']:.2f}")
            print(f"Order Status  : {order['order_status']}")
            print("-" * 35)

    except FileNotFoundError:
        print("\nNo order history found. orders.json does not exist yet.")


# Main program
while True:
    print("\n===== Food Delivery Order Manager =====")
    print("1. Add New Order")
    print("2. View Orders")
    print("0. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_order()

    elif choice == "2":
        view_orders()

    elif choice == "0":
        print("Thank you! Program ended.")
        break

    else:
        print("Invalid choice. Please try again.")
```
