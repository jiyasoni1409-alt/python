```python
import csv
import os


# Rider Class
class Rider:
    def __init__(self, rider_id, name, status="Available", total_deliveries=0):
        self.rider_id = rider_id
        self.name = name
        self.status = status
        self.total_deliveries = total_deliveries

    # Assign an order
    def assign_order(self, order_id):
        self.status = "On Delivery"
        print(f"Order {order_id} assigned to Rider {self.rider_id} - {self.name}")

    # Complete delivery
    def complete_delivery(self):
        self.total_deliveries += 1
        self.status = "Available"
        print(f"Delivery completed by Rider {self.rider_id} - {self.name}")

    # Display rider information
    def display_info(self):
        print("Rider ID:", self.rider_id)
        print("Name:", self.name)
        print("Status:", self.status)
        print("Total Deliveries:", self.total_deliveries)
        print("-" * 30)


# Save riders to CSV file
def save_riders(riders):
    with open("riders.csv", "w", newline="") as file:
        writer = csv.writer(file)

        writer.writerow([
            "rider_id",
            "name",
            "status",
            "total_deliveries"
        ])

        for rider in riders:
            writer.writerow([
                rider.rider_id,
                rider.name,
                rider.status,
                rider.total_deliveries
            ])

    print("\nRider data saved successfully to riders.csv")


# Load riders from CSV file
def load_riders():
    riders = []

    if os.path.exists("riders.csv"):
        with open("riders.csv", "r", newline="") as file:
            reader = csv.DictReader(file)

            for row in reader:
                rider = Rider(
                    int(row["rider_id"]),
                    row["name"],
                    row["status"],
                    int(row["total_deliveries"])
                )
                riders.append(rider)

        print("Previous rider data loaded from riders.csv")

    else:
        riders = [
            Rider(1, "Rahul"),
            Rider(2, "Amit"),
            Rider(3, "Priya")
        ]

        print("No previous file found. Three new riders created.")

    return riders


# Display menu
def show_menu():
    print("\n========== DELIVERY RIDER SYSTEM ==========")
    print("1. Display all riders")
    print("2. Assign order")
    print("3. Complete delivery")
    print("4. Exit")
    print("===========================================")


# Main Program
riders = load_riders()

while True:

    show_menu()

    choice = input("Enter your choice: ")

    # Option 1: Display riders
    if choice == "1":

        print("\n----- Rider Details -----")

        for rider in riders:
            rider.display_info()

    # Option 2: Assign order
    elif choice == "2":

        rider_id = int(input("Enter Rider ID: "))
        order_id = input("Enter Order ID: ")

        found = False

        for rider in riders:
            if rider.rider_id == rider_id:
                rider.assign_order(order_id)
                found = True
                break

        if not found:
            print("Rider not found.")

    # Option 3: Complete delivery
    elif choice == "3":

        rider_id = int(input("Enter Rider ID: "))

        found = False

        for rider in riders:
            if rider.rider_id == rider_id:
                rider.complete_delivery()
                found = True
                break

        if not found:
            print("Rider not found.")

    # Option 4: Exit
    elif choice == "4":

        save_riders(riders)
        print("Thank you for using the Delivery Rider System!")
        break

    # Invalid choice
    else:
        print("Invalid choice. Please enter 1, 2, 3, or 4.")
```
