Write a recursive function in Python called reverse_string(s) that takes a string and returns it reversed (e.g., 'hello' becomes 'olleh').

```python
def reverse_string(s):

    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]


print(reverse_string("hello"))
```

2.
Build a recursive function sum_playlist_durations(durations) that takes a list of song durations (in seconds) and returns the total duration, similar to how Spotify totals a playlist.

```python
def sum_playlist_durations(durations):
    # Base case
    if len(durations) == 0:
        return 0
    
    # Add first duration and call function for remaining durations
    return durations[0] + sum_playlist_durations(durations[1:])


playlist = [180, 240, 200, 300]

print("Total duration:", sum_playlist_durations(playlist), "seconds")
```

```python
# Global variable
app_status = "Offline"

print("Before function call:")
print("App status:", app_status)


def update_status():
    # Local variable
    user_status = "Online"

    print("\nDuring function call:")
    print("User status:", user_status)
    print("App status:", app_status)


update_status()

print("\nAfter function call:")
print("App status:", app_status)
```
