3) .Write a function remove_last_item(order_list) that pops the last item from a Zomato order list and returns the removed item. Test it with a sample order_list.

```python
def lastitem(order_list):
    removeditem = order_list.pop()
    return removeditem


order_list = ['pizza', 'burger', 'cold drink']

removed = lastitem(order_list)

print("Removed item:", removed)
print("Updated order:", order_list)
```

```python
insta_filters = ('Clarendon', 'Juno', 'Lark', 'Gingham')

try:
    insta_filters[1] = 'Valencia'
except TypeError as e:
    print("Error:", e)
# Error occurs because tuples are immutable.
# We cannot change, add, or remove  elements of a tuple.
```
