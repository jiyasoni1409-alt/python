2.
In your script, measure and print the memory usage (in bytes) of both the Python list and the NumPy array containing 1000 integers.<br><br><em><strong>Hint:</strong> Use the sys.getsizeof() function for the list and the nbytes attribute for the NumPy array.</em>


```python
import numpy as np
import sys

# Create Python list
my_list = list(range(1, 1001))

# Create NumPy array
my_array = np.arange(1, 1001)

# Calculate memory usage
list_memory = sys.getsizeof(my_list)
array_memory = my_array.nbytes

# Print memory usage
print("Memory usage of Python List:", list_memory, "bytes")
print("Memory usage of NumPy Array:", array_memory, "bytes")
```

3.
Write a function compare_addition_speed() that adds 5 to every element in both a Python list and a NumPy array of 10,000 integers, and prints the time taken for each.<br><br><em><strong>Hint:</strong> Use the time module to measure execution time.</em>


```python
import numpy as np
import time

def compare_addition_speed():

    # Create Python list
    my_list = list(range(1, 10001))

    # Create NumPy array
    my_array = np.arange(1, 10001)

    # Python list addition using for-loop
    start_time = time.time()

    list_result = []
    for number in my_list:
        list_result.append(number + 5)

    list_time = time.time() - start_time

    start_time = time.time()

    array_result = my_array + 5

    array_time = time.time() - start_time

    print("Time taken by Python List:", list_time, "seconds")
    print("Time taken by NumPy Array:", array_time, "seconds")


compare_addition_speed()
```

Explain with code how vectorized operations in NumPy can replace for-loops when multiplying all elements of an array by 2. Show both the loop and the vectorized version using a Zomato-style example: multiplying all restaurant ratings by 2.

```python
import numpy as np


ratings = np.array([2.5, 3.0, 3.5, 4.0, 4.5])

# Vectorized operation
new_ratings = ratings * 2

print("Original Ratings:")
print(ratings)

print("\nRatings after multiplying by 2:")
print(new_ratings)
```

using for loop


```python
import numpy as np

# Zomato-style restaurant ratings
ratings = np.array([2.5, 3.0, 3.5, 4.0, 4.5])

# Empty list to store results
new_ratings = []

# Multiply each rating by 2 using a loop
for rating in ratings:
    new_ratings.append(rating * 2)

print("Original Ratings:")
print(ratings)

print("\nRatings after multiplying by 2:")
print(new_ratings)
```
