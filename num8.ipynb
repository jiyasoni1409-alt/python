1.
Given a NumPy array of IPL team names with some duplicates, use np.unique() to print a sorted list of all unique team names.


```python
import numpy as np

teams = np.array([
    "MI", "CSK", "RCB", "MI", "KKR",
    "CSK", "GT", "RCB", "MI", "GT"
])

unique_teams = np.unique(teams)

print("Original teams:")
print(teams)

print("\nUnique IPL teams:")
print(unique_teams)
```

2.
Create a NumPy array of Zomato order ratings (with some NaN values), then use np.isnan() to count how many ratings are missing.


```python
import numpy as np

ratings = np.array([4.5, np.nan, 3.8, 4.2, np.nan, 5.0, 3.5])

missing_values = np.isnan(ratings)

missing_count = np.sum(missing_values)

print("Zomato ratings:")
print(ratings)

print("\nMissing value positions:")
print(missing_values)

print("\nNumber of missing ratings:", missing_count)
```

3.
Given an array of Flipkart product prices, use np.clip() to limit all prices between 100 and 1000, and print the resulting array.<br><br><em><strong>Hint:</strong> Use np.clip(array, 100, 1000).</em>


```python
import numpy as np

prices = np.array([50, 150, 450, 750, 1200, 2000, 80, 999])

clipped_prices = np.clip(prices, 100, 1000)

print("Original prices:")
print(prices)

print("\nPrices after clipping:")
print(clipped_prices)
```

4.
You have a NumPy array of YouTube video view counts, some of which are NaN or inf. Replace all NaN values with 0, and all inf values with the maximum finite value in the array.


```python
import numpy as np

views = np.array([
    1000, 2500, np.nan, 5000,
    np.inf, 7500, np.nan, 10000
])

print("Original views:")
print(views)

max_finite = np.max(views[np.isfinite(views)])

clean_views = np.nan_to_num(
    views,
    nan=0,
    posinf=max_finite,
    neginf=max_finite
)

print("\nMaximum finite value:", max_finite)

print("\nCleaned views:")
print(clean_views)
```

5.
Use ChatGPT to generate a Python code snippet that finds the indices of all even numbers in a NumPy array using np.where(), then test the code with your own example array.

```python
import numpy as np

numbers = np.array([11, 24, 35, 42, 57, 68, 73, 80, 91, 100])

even_indices = np.where(numbers % 2 == 0)

print("Original array:")
print(numbers)

print("\nIndices of even numbers:")
print(even_indices[0])
```
